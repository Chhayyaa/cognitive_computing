Q1

```python
import pandas as pd

roll_number = input("Enter your college roll number: ").strip()
last_3_digits = roll_number[-3:]

categories = ["billing", "account", "general"]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

constructed_entries = []

for digit in last_3_digits:
    category = categories[int(digit) % 3]
    keyword = {
        "billing": "fee payment bill",
        "account": "password login account",
        "general": "timing information help"
    }[category]

    question = {
        "billing": "how do i pay my bill",
        "account": "how do i update my account",
        "general": "what information do you provide"
    }[category]

    answer = {
        "billing": "You can pay your bill using the available payment methods.",
        "account": "You can update your account details from your account settings.",
        "general": "We provide general information and assistance."
    }[category]

    constructed_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keyword,
        "category": category
    })

faq_data = fixed_entries + constructed_entries
df = pd.DataFrame(faq_data)

df
```

Q2

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        matches = query_words.intersection(keywords)
        score = len(matches) / len(keywords) if keywords else 0

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "keywords": row["keywords"],
                "category": row["category"],
                "score": score
            })

    return pd.DataFrame(results).sort_values(
        by="score", ascending=False
    ).reset_index(drop=True)

query = input("Enter your query: ")
score_query(query, df)
```

```python
def score_query_with_ties(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        matches = query_words.intersection(keywords)
        score = len(matches) / len(keywords) if keywords else 0

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "keywords": row["keywords"],
                "category": row["category"],
                "score": score
            })

    if not results:
        return pd.DataFrame(
            columns=["question", "answer", "keywords", "category", "score"]
        )

    result_df = pd.DataFrame(results).sort_values(
        by="score", ascending=False
    ).reset_index(drop=True)

    highest_score = result_df["score"].max()
    return result_df[result_df["score"] == highest_score].reset_index(drop=True)

query = input("Enter a query to demonstrate matching ties: ")
score_query_with_ties(query, df)
```
