```python
emp_df = pd.read_csv("employees.csv")
emp_df
```

```python
print("First 5 rows:")
display(emp_df.head(5))

print("Last 3 rows:")
display(emp_df.tail(3))
```

```python
average_salary = emp_df["Salary"].mean()
print("Average salary:", average_salary)
```

```python
total_bonus = emp_df["Bonus"].sum()
print("Total bonus:", total_bonus)
```

```python
youngest_age = emp_df["Age"].min()
print("Youngest age:", youngest_age)
```

```python
highest_rating = emp_df["Rating"].max()
print("Highest rating:", highest_rating)
```

```python
salary_sorted = emp_df.sort_values(by="Salary", ascending=False)
salary_sorted
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance"] = emp_df["Rating"].apply(performance_category)
emp_df
```

```python
print("Missing values in each column:")
print(emp_df.isnull().sum())

print("\nDoes the DataFrame contain any missing value?")
print(emp_df.isnull().values.any())
```

```python
emp_df = emp_df.rename(columns={"Employee_ID": "ID"})
emp_df
```

```python
more_than_5_years = emp_df[emp_df["Years_of_Experience"] > 5]
more_than_5_years
```

```python
it_employees = emp_df[emp_df["Department"] == "IT"]
it_employees
```

```python
emp_df["Tax"] = emp_df["Salary"] * 0.10
emp_df
```

```python
output_file = "modified_employees.csv"
emp_df.to_csv(output_file, index=False)

print(f"Modified DataFrame saved as: {output_file}")
```
