```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temperature_error = temperature + 2
fahrenheit = (9/5) * temperature + 32
above_32 = temperature[temperature > 32]
count_above_32 = np.sum(temperature > 32)

print("Corrected temperatures:", temperature_error)
print("Fahrenheit temperatures:", fahrenheit)
print("Readings greater than 32°C:", above_32)
print("Number of readings exceeding 32°C:", count_above_32)
print("Vectorization applies operations to an entire NumPy array at once.")
print("Boolean indexing directly selects elements satisfying a condition.")
```

Q2

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = np.sum(steps)
mean_steps = np.mean(steps)
maximum_steps = np.max(steps)
minimum_steps = np.min(steps)
total_each_day = np.sum(steps, axis=0)
total_each_user = np.sum(steps, axis=1)
maximum_position = np.unravel_index(np.argmax(steps), steps.shape)

print("Total steps:", total_steps)
print("Mean steps:", mean_steps)
print("Maximum steps:", maximum_steps)
print("Minimum steps:", minimum_steps)
print("Total steps for each day:", total_each_day)
print("Total steps for each user:", total_each_user)
print("Position of maximum steps (user, day):", maximum_position)
```

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500

print("Original after modifying copied array:", original)
print("Copied array:", copied)

matrix = np.arange(1, 13).reshape(3, 4)

print("Matrix:")
print(matrix)
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1–2 and columns 2–3:")
print(matrix[0:2, 1:3])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("Flattened array:", flattened)
print("Raveled array:", raveled)

raveled[0] = 100
print("Matrix after modifying ravel():")
print(matrix)

flattened[1] = 200
print("Matrix after modifying flatten():")
print(matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("X.T:")
print(XT)

print("Transpose represents features as rows and users as columns.")

XTX = X.T @ X
print("X.T @ X:")
print(XTX)

XTX_inv = np.linalg.inv(XTX)
print("Inverse of X.T @ X:")
print(XTX_inv)

beta = XTX_inv @ X.T @ y
print("OLS coefficients:")
print(beta)

print("The coefficients represent the change in predicted assistance score associated with a one-unit change in sleep hours, activity level, and stress level, respectively.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
