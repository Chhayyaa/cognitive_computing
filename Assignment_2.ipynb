## Question 3 

```python
import random
from collections import Counter

random.seed(1024170343)

numbers = [random.randint(100, 900) for _ in range(100)]

print("100 random numbers:")
print(numbers)
```

### Q3(i) Generate 100 random numbers

```python
print(numbers)
```

### Q3(ii) Count all numbers

```python
print("Number of elements:", len(numbers))
```

### Q3(iii) Even numbers

```python
even_numbers = [x for x in numbers if x % 2 == 0]

print("Even numbers:")
print(even_numbers)
print("Count:", len(even_numbers))
```

### Q3(iv) Prime numbers

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime numbers:")
print(prime_numbers)
print("Count:", len(prime_numbers))
```

### Q3(v) Most frequent value

```python
frequency = Counter(numbers)
max_frequency = max(frequency.values())

most_frequent = [
    x for x, count in frequency.items()
    if count == max_frequency
]

print("Most frequent value(s):", most_frequent)
print("Number of occurrences:", max_frequency)
```

### Q5(iv) Delete the `branch` key using `pop()` and `del`

```python

dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")

print("After pop():")
print(dict_pop)
print("Value returned by pop():", removed_branch)

dict_del = my_dict.copy()
del dict_del["branch"]

print("\nAfter del:")
print(dict_del)
```

### Q5(vii) Merge with a friend dictionary using `{**dict1, **dict2}`


```python

friend_dict = {
    "name": "Friend",
    "roll_no": "1024170344",
    "branch": "ECE",
    "age": 19,
    "city": "Delhi"
}

dict1 = {
    "name": my_dict["name"],
    "roll_no": my_dict["roll_no"],
    "branch": "CSE",
    "age": my_dict["age"],
    "city": "Patiala"
}

dict2 = friend_dict.copy()

merged_dict = {**dict1, **dict2}

print("Dictionary 1:")
print(dict1)

print("\nDictionary 2:")
print(dict2)

print("\nMerged dictionary:")
print(merged_dict)

print("\nWhen both dictionaries share a key, the value from dict2 wins.")
```

### Q5(viii) Dictionary comprehension – only string values

```python

string_values_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Dictionary containing only string values:")
print(string_values_dict)
```
