```python
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 200)
y1 = M * x**2
y2 = M * np.sin(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y1, color="blue", linestyle="-", label="y = Mx²")
plt.plot(x, y2, color="red", linestyle="--", label="y = M sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Mathematical Functions")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

subjects = ["Mathematics", "Physics", "Chemistry", "Computer Science", "English"]
scores = [85, 78, 92, 88, 81]

df = pd.DataFrame({"Subject": subjects, "Score": scores})

plt.figure(figsize=(10, 6))
ax = sns.barplot(data=df, x="Subject", y="Score", hue="Subject", palette="Set2", legend=False)

for i, score in enumerate(scores):
    ax.text(i, score + 1, str(score), ha="center", va="bottom")

plt.title("Subject Scores")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

roll_number = int(input("Enter your roll number: "))
np.random.seed(roll_number)

data = np.random.randn(50)
cumulative_sum = np.cumsum(data)
noise = np.random.normal(0, 0.5, 50)
x = np.arange(1, 51)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].plot(x, cumulative_sum, linewidth=2)
axes[0, 0].set_title("Cumulative Sum")
axes[0, 0].set_xlabel("Index")
axes[0, 0].set_ylabel("Cumulative Sum")
axes[0, 0].grid(True)

axes[0, 1].scatter(x, noise, alpha=0.7)
axes[0, 1].set_title("Random Noise")
axes[0, 1].set_xlabel("Index")
axes[0, 1].set_ylabel("Noise")
axes[0, 1].grid(True)

axes[1, 0].plot(x, data, linewidth=1.5)
axes[1, 0].set_title("Random Data")
axes[1, 0].set_xlabel("Index")
axes[1, 0].set_ylabel("Value")
axes[1, 0].grid(True)

axes[1, 1].scatter(data, noise, alpha=0.7)
axes[1, 1].set_title("Data vs Noise")
axes[1, 1].set_xlabel("Random Data")
axes[1, 1].set_ylabel("Noise")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
df = pd.read_csv(url)

df.head()
```

```python
month_col = next(col for col in df.columns if "month" in col.lower())
profit_col = next(col for col in df.columns if "profit" in col.lower())

plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x=month_col, y=profit_col, marker="o")
plt.title("Total Profit of All Months")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()
```

```python
sales_cols = [col for col in df.columns if "sales" in col.lower()]

plt.figure(figsize=(12, 7))
for col in sales_cols:
    sns.lineplot(data=df, x=month_col, y=col, marker="o", label=col)

plt.title("Product Sales by Month")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

fig, axes = plt.subplots(len(numeric_cols), 1, figsize=(12, 5 * len(numeric_cols)))

if len(numeric_cols) == 1:
    axes = [axes]

for ax, col in zip(axes, numeric_cols):
    sns.barplot(data=df, x=month_col, y=col, ax=ax)
    ax.set_title(f"{col} by Month")
    ax.set_xlabel("Month")
    ax.set_ylabel(col)
    ax.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```
