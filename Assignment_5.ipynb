## Q1. Create a NumPy 1D array with 5 elements and perform basic operations

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print("Original array:", arr)

add_result = arr + 2
print("After adding 2 to all elements:", add_result)

mul_result = arr * 3
print("After multiplying 3 with all elements:", mul_result)

div_result = arr / 2
print("After dividing every element by 2:", div_result)
```

## Q2. Questions on Basic NumPy Array

### a) Reverse the NumPy array

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Original array:", arr)
print("Reversed array:", reversed_arr)
```

### b) Find the most frequent value and their indices

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent_x = values[np.argmax(counts)]
indices_x = np.where(x == most_frequent_x)[0]

print("Array x:", x)
print("Most frequent value in x:", most_frequent_x)
print("Indices of most frequent value in x:", indices_x)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent_y = values[np.argmax(counts)]
indices_y = np.where(y == most_frequent_y)[0]

print("Array y:", y)
print("Most frequent value in y:", most_frequent_y)
print("Indices of most frequent value in y:", indices_y)
```

## Q3. Accessing elements in a 2-D array using row and column indices

```python
arr = np.array([[10, 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]])
print("2-D array:\n", arr)

a_val = arr[0, 1]
print("Element at 1st row, 2nd column:", a_val)

b_val = arr[2, 0]
print("Element at 3rd row, 1st column:", b_val)
```

## Q4. 1-D NumPy array with evenly spaced numbers using linspace()

```python
Smarth = np.linspace(10, 100, 25)
print("Array:", Smarth)

print("\nDimensions (ndim):", Smarth.ndim)
print("Shape:", Smarth.shape)
print("Total elements (size):", Smarth.size)
print("Data type of each element (dtype):", Smarth.dtype)
print("Total bytes consumed (nbytes):", Smarth.nbytes)
```

### Transpose using reshape()
Since the array is 1-D, `reshape()` can be used to change its shape (for example into a column vector), which acts like a transpose for a 1-D array.

```python
transposed_reshape = Your_Name.reshape(25, 1)
print("Shape after reshape(25, 1):", transposed_reshape.shape)
print(transposed_reshape)
```

### Can we do the same with the T attribute?
No — for a 1-D array, the `.T` attribute has **no effect**. Transpose only swaps axes, and a 1-D array has just one axis, so `.T` returns the array unchanged (still shape `(25,)`). To truly get a transposed 2-D (row/column) effect, the array must first be reshaped into 2-D (e.g., using `reshape(25, 1)` or `reshape(1, 25)`), after which `.T` will work as expected.

```python
print("Original shape:", Your_Name.shape)
print("Shape after .T on 1-D array:", Your_Name.T.shape)

reshaped_2d = Your_Name.reshape(1, 25)
print("\nShape of 2-D version:", reshaped_2d.shape)
print("Shape after applying .T on 2-D version:", reshaped_2d.T.shape)
```

## Q5. 2-D Array operations - mean, median, max, min, unique, reshape, resize

```python
ucs420_Your_Name = np.array([[10, 20, 30, 40],
                              [50, 60, 70, 80],
                              [90, 15, 20, 35]])
print("Original array (3x4):\n", ucs420_Your_Name)

mean_val = np.mean(ucs420_Your_Name)
median_val = np.median(ucs420_Your_Name)
max_val = np.max(ucs420_Your_Name)
min_val = np.min(ucs420_Your_Name)
unique_vals = np.unique(ucs420_Your_Name)

print("\nMean:", mean_val)
print("Median:", median_val)
print("Max:", max_val)
print("Min:", min_val)
print("Unique elements:", unique_vals)
```

```python
reshaped_ucs420_Your_Name = ucs420_Your_Name.reshape(4, 3)
print("Reshaped array (4x3):\n", reshaped_ucs420_Your_Name)
```

```python
resized_ucs420_Your_Name = np.resize(ucs420_Your_Name, (2, 3))
print("Resized array (2x3):\n", resized_ucs420_Your_Name)
```
